```python
# Load the credit card dataset
creditcard_data = pd.read_csv('../data/creditcard.csv')

# Load the e-commerce fraud dataset
fraud_data = pd.read_csv('../data/Fraud_Data_Processed.csv')  # Ensure this is the preprocessed dataset

# Display the first few rows of each dataset
print("Credit Card Dataset:")
display(creditcard_data.head())

print("\nE-Commerce Fraud Dataset:")
display(fraud_data.head())
```

```python
# Separate features and target for the credit card dataset
X_creditcard = creditcard_data.drop(columns=['Class'])
y_creditcard = creditcard_data['Class']
```

```python
# Separate features and target for the e-commerce fraud dataset
X_fraud = fraud_data.drop(columns=['class'])
y_fraud = fraud_data['class']
```

```python
from sklearn.model_selection import train_test_split

# Split the credit card dataset
X_train_creditcard, X_test_creditcard, y_train_creditcard, y_test_creditcard = train_test_split(
    X_creditcard, y_creditcard, test_size=0.2, random_state=42, stratify=y_creditcard
)

# Split the e-commerce fraud dataset
X_train_fraud, X_test_fraud, y_train_fraud, y_test_fraud = train_test_split(
    X_fraud, y_fraud, test_size=0.2, random_state=42, stratify=y_fraud
)
```

```python
# Check class distribution in the credit card dataset
print("Credit Card Dataset - Training Set Class Distribution:")
print(y_train_creditcard.value_counts(normalize=True))

print("\nCredit Card Dataset - Testing Set Class Distribution:")
print(y_test_creditcard.value_counts(normalize=True))

# Check class distribution in the e-commerce fraud dataset
print("\nE-Commerce Fraud Dataset - Training Set Class Distribution:")
print(y_train_fraud.value_counts(normalize=True))

print("\nE-Commerce Fraud Dataset - Testing Set Class Distribution:")
print(y_test_fraud.value_counts(normalize=True))
```
